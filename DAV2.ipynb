```python
import numpy as np
import pandas as pd
print("Numpy version:",np.__version__)
print("Pandas version:",pd.__version__)
```

```python
Marks= np.array([43,56,76,89])
print("Type:",type(Marks))
```

```python
print("Original:",Marks)
print("Add 5:",Marks+5)
print("Subtract 5:",Marks-5)
print("Multiply 5:",Marks*5)
print("Divide 5:",Marks/5)
```

```python
print("Sum:",Marks.sum())
print("Min:",Marks.min())
print("Max:",Marks.max())
print("Mean:",Marks.mean())
```

```python
print("Index 1 to 3:",Marks[1:4])
print("First three",Marks[:3])
print("From index 2:",Marks[2:])
print("Every second element:",Marks[::2])
```

```python
DATA = {
    "Name":["Sharat","Ridam","Sanjay","Puneeth","Alex"],
    "Branch":["CSE","ECE","MECH","AIML","CIVIL"],
    "Marks":[88,55,86,78,75],
    "Attendence":[85,86,88,84,87]
}
std = pd.DataFrame(DATA)
std
```

```python
print("display Name:")
display(std.Name)
```

```python
print("Display students who scored more than 80:")
scores = std[std["Marks"]>80]
scores
```

```python
attd=std[std["Attendence"]>85]
attd
```

```python
print("name of first three students")
display(std[:3])
```

```python
print("Average marks:")
avg=np.mean(std["Marks"])
print(avg)
```

```python
print("Maximum marks:")
print(np.max(std["Marks"]))
```

```python
print("Students from CSE branch:")
display(std[std["Branch"]=="CSE"])
```

```python
print("Student scored more than 80 and attendence more than 85:")
display(std[(std["Marks"]>80) & (std["Attendence"]>85)])
```

```python
print("Display Name and Marks:")
col=["Name","Marks"]
display(std[col])
```
